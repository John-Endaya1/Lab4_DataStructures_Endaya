```python
# CELL 1 — STUDENT-SPECIFIC INPUTS
LAST_NAME = "ENDAYA"
FAVORITE_ARTIST = "ChaseAtlantic"
SEED_NUM = 4330
SEED_DIGIT = 0
ID_SUM = 18

print("STUDENT INPUTS")
print("LAST_NAME      :", LAST_NAME)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_NUM       :", SEED_NUM)
print("SEED_DIGIT     :", SEED_DIGIT)
print("ID_SUM         :", ID_SUM)
```

```python
# CELL 2 — GENERATE ENGINEERING TEST RECORDS
from collections import Counter, defaultdict
from statistics import mean, median

test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Use character positions (1-based) from FAVORITE_ARTIST to generate reproducible values.
artist_positions = [ord(ch) for ch in FAVORITE_ARTIST if ch.isalnum()]
if not artist_positions:
    artist_positions = [1]

test_records = []
for i, source in enumerate(test_sources):
    for j, category in enumerate(categories):
        char_value = artist_positions[(i + j) % len(artist_positions)]
        value = round(((SEED_NUM % 100) + char_value + (i + 1) * 3 + (j + 1) * 2) / 10, 2)
        test_records.append({"source": source, "category": category, "value": value})

# Required repeated source/category record
test_records.append(test_records[0].copy())

# Additional measurements under at least two categories
test_records.extend([
    {"source": test_sources[1], "category": "TEMPERATURE", "value": round(test_records[6]["value"] + 1.5, 2)},
    {"source": test_sources[3], "category": "VOLTAGE", "value": round(test_records[16]["value"] + 2.0, 2)}
])

print(f"Generated records: {len(test_records)}")
for record in test_records:
    print(record)
```

```python
# CELL 3 — ORGANIZE RECORDS BY SOURCE AND CATEGORY
organized_tests = defaultdict(lambda: defaultdict(list))
for record in test_records:
    organized_tests[record["source"]][record["category"]].append(record["value"])

for source in sorted(organized_tests):
    print(f"\n{source}")
    for category in sorted(organized_tests[source]):
        print(f"  {category}: {organized_tests[source][category]}")
```

```python
# CELL 4 — REPEATED/DISTINCT DATA AND SUMMARIES
record_keys = [(r["source"], r["category"], r["value"]) for r in test_records]
key_counts = Counter(record_keys)
repeated_records = {key: count for key, count in key_counts.items() if count > 1}
distinct_records = len(set(record_keys))

all_values = [r["value"] for r in test_records]
category_values = defaultdict(list)
for record in test_records:
    category_values[record["category"]].append(record["value"])

print("Repeated exact records:")
for key, count in repeated_records.items():
    print(f"  {key} occurred {count} times")
print("Total records:", len(test_records))
print("Distinct exact records:", distinct_records)
print("Numerical summary:")
print(f"  Count={len(all_values)}, Min={min(all_values):.2f}, Max={max(all_values):.2f}, Mean={mean(all_values):.2f}, Median={median(all_values):.2f}")
print("\nCategorical summary:")
for category in categories:
    vals = category_values[category]
    print(f"  {category}: count={len(vals)}, mean={mean(vals):.2f}, min={min(vals):.2f}, max={max(vals):.2f}")
```

```python
# CELL 5 — ORDERED RESULTS AND FINAL SUMMARY
ordered_tests = sorted(test_records, key=lambda r: (r["source"], r["category"], r["value"]))
print("ORDERED TEST RECORDS")
for record in ordered_tests:
    print(f'{record["source"]:18} | {record["category"]:12} | {record["value"]:7.2f}')

print("\nFINAL ENGINEERING TEST DATA SUMMARY")
print(f"Sources: {len(test_sources)}")
print(f"Categories: {len(categories)}")
print(f"Total measurements: {len(test_records)}")
print(f"Repeated exact record types: {len(repeated_records)}")
print(f"Distinct exact records: {distinct_records}")
print(f"Overall average measurement: {mean(all_values):.2f}")
```

```python
# CELL 6 — GENERATE STUDENT PERFORMANCE RECORDS
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "EXCELLENT"
    if score >= 80:
        return "GOOD"
    if score >= 70:
        return "SATISFACTORY"
    return "NEEDS IMPROVEMENT"

student_records = []
for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        char_value = artist_positions[(i + j) % len(artist_positions)]
        score = 60 + ((SEED_NUM + char_value + (i + 1) * 7 + (j + 1) * 5) % 41)
        student_records.append({
            "student": student,
            "subject": subject,
            "score": score,
            "category": performance_category(score)
        })

# Required repeated student-subject record
student_records.append(student_records[0].copy())

# Required incomplete record (missing score)
student_records.append({
    "student": students[2],
    "subject": "PHYSICS",
    "score": None,
    "category": "INCOMPLETE"
})

print(f"Generated student performance records: {len(student_records)}")
for record in student_records:
    print(record)
```

```python
# CELL 7 — ORGANIZE, IDENTIFY DUPLICATES, AND INCOMPLETE RECORDS
student_subject_data = defaultdict(dict)
for record in student_records:
    student_subject_data[record["student"]][record["subject"]] = record["score"]

student_record_keys = [(r["student"], r["subject"], r["score"]) for r in student_records]
student_counts = Counter(student_record_keys)
repeated_student_records = {k: v for k, v in student_counts.items() if v > 1}
incomplete_records = [r for r in student_records if r["score"] is None]

print("ORGANIZED STUDENT/SUBJECT DATA")
for student in sorted(student_subject_data):
    print(f"{student}: {student_subject_data[student]}")

print("\nRepeated records:")
for record, count in repeated_student_records.items():
    print(f"  {record} occurred {count} times")
print("\nIncomplete records:")
for record in incomplete_records:
    print(" ", record)
```

```python
# CELL 8 — PERFORMANCE CATEGORIES AND SUMMARIES
valid_records = [r for r in student_records if r["score"] is not None]
scores_by_student = defaultdict(list)
scores_by_subject = defaultdict(list)
for record in valid_records:
    scores_by_student[record["student"]].append(record["score"])
    scores_by_subject[record["subject"]].append(record["score"])

print("CATEGORY COUNTS")
category_counts = Counter(r["category"] for r in student_records)
for category, count in sorted(category_counts.items()):
    print(f"  {category}: {count}")

print("\nSTUDENT SUMMARIES (valid scores only)")
for student in students:
    vals = scores_by_student[student]
    print(f"  {student}: count={len(vals)}, average={mean(vals):.2f}, highest={max(vals)}, lowest={min(vals)}")

print("\nSUBJECT SUMMARIES (valid scores only)")
for subject in subjects:
    vals = scores_by_subject[subject]
    print(f"  {subject}: count={len(vals)}, average={mean(vals):.2f}, highest={max(vals)}, lowest={min(vals)}")
```

```python
# CELL 9 — ORDERED RESULTS AND FINAL STUDENT PERFORMANCE REPORT
ordered_student_records = sorted(
    student_records,
    key=lambda r: (r["student"], r["subject"], -1 if r["score"] is None else r["score"])
)

print("ORDERED PERFORMANCE RECORDS")
for r in ordered_student_records:
    score_text = "MISSING" if r["score"] is None else str(r["score"])
    print(f'{r["student"]:16} | {r["subject"]:13} | {score_text:7} | {r["category"]}')

print("\nFINAL STUDENT PERFORMANCE REPORT")
print("Total records:", len(student_records))
print("Valid scores:", len(valid_records))
print("Incomplete records:", len(incomplete_records))
print("Repeated record types:", len(repeated_student_records))
print(f"Overall average of valid scores: {mean([r['score'] for r in valid_records]):.2f}")
print("Performance category totals:", dict(category_counts))
```

```python
# CELL 10 — GENERATE TRANSACTION RECORDS
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {
    "NOTEBOOK": 45.00,
    "PEN": 15.00,
    "USB": 350.00,
    "HEADPHONES": 650.00,
    "CALCULATOR": 250.00
}

transactions = []
for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        char_value = artist_positions[(i + j) % len(artist_positions)]
        quantity = 1 + ((SEED_NUM + char_value + i + j) % 5)
        transactions.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "unit_price": prices[product],
            "total": round(quantity * prices[product], 2)
        })

# Required repeated customer-product purchase
transactions.append(transactions[0].copy())

print("GENERATED TRANSACTION RECORDS")
for t in transactions:
    print(t)
```

```python
# CELL 11 — ORGANIZE TRANSACTIONS AND IDENTIFY REPEATED/DISTINCT PURCHASES
transaction_keys = [(t["customer"], t["product"], t["quantity"], t["unit_price"]) for t in transactions]
transaction_counts = Counter(transaction_keys)
repeated_purchases = {k: count for k, count in transaction_counts.items() if count > 1}
distinct_transactions = len(set(transaction_keys))

transactions_by_customer = defaultdict(list)
transactions_by_product = defaultdict(list)
for t in transactions:
    transactions_by_customer[t["customer"]].append(t)
    transactions_by_product[t["product"]].append(t)

print("Repeated customer-product transaction(s):")
for key, count in repeated_purchases.items():
    print(f"  {key} occurred {count} times")
print("Total transaction records:", len(transactions))
print("Distinct transaction records:", distinct_transactions)
print("\nTransactions organized by customer:")
for customer in customers:
    print(f"  {customer}: {len(transactions_by_customer[customer])} records")
print("\nTransactions organized by product:")
for product in products:
    print(f"  {product}: {len(transactions_by_product[product])} records")
```

```python
# CELL 12 — TRANSACTION VALUES AND CUSTOMER/PRODUCT SUMMARIES
overall_value = sum(t["total"] for t in transactions)

print("CUSTOMER SUMMARIES")
for customer in customers:
    records = transactions_by_customer[customer]
    total_spent = sum(t["total"] for t in records)
    total_units = sum(t["quantity"] for t in records)
    print(f"  {customer}: transactions={len(records)}, units={total_units}, total_spent=₱{total_spent:,.2f}")

print("\nPRODUCT SUMMARIES")
for product in products:
    records = transactions_by_product[product]
    total_units = sum(t["quantity"] for t in records)
    total_sales = sum(t["total"] for t in records)
    print(f"  {product}: unit_price=₱{prices[product]:,.2f}, units_sold={total_units}, sales=₱{total_sales:,.2f}")

print("\nOVERALL TRANSACTION SUMMARY")
print("Total transaction records:", len(transactions))
print("Distinct transaction records:", distinct_transactions)
print("Total units purchased:", sum(t["quantity"] for t in transactions))
print(f"Overall transaction value: ₱{overall_value:,.2f}")
```

```python
# CELL 13 — ORDERED RESULTS AND FINAL CONSOLIDATED TRANSACTION REPORT
ordered_transactions = sorted(
    transactions,
    key=lambda t: (t["customer"], t["product"], t["total"])
)

print("ORDERED TRANSACTION RECORDS")
for t in ordered_transactions:
    print(
        f'{t["customer"]:16} | {t["product"]:12} | '
        f'Qty={t["quantity"]} | Unit=₱{t["unit_price"]:,.2f} | Total=₱{t["total"]:,.2f}'
    )

print("\nFINAL CONSOLIDATED TRANSACTION REPORT")
print(f"Customers: {len(customers)}")
print(f"Products: {len(products)}")
print(f"Transaction records: {len(transactions)}")
print(f"Repeated purchase record types: {len(repeated_purchases)}")
print(f"Distinct transaction records: {distinct_transactions}")
print(f"Total units purchased: {sum(t['quantity'] for t in transactions)}")
print(f"Total transaction value: ₱{overall_value:,.2f}")
print("Product with highest sales:", max(products, key=lambda p: sum(t["total"] for t in transactions_by_product[p])))
```
